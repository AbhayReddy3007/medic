```python
from google import genai
from google.genai import types

# Initialize the standard client
client = genai.Client()

try:
    # Ask a question and enable Google Search grounding
    response = client.models.generate_content(
        model="gemini-3.5-flash-lite",
        contents="What are the latest major developments in GLP-1 medicines? do a web search think and respond properly",
        config=types.GenerateContentConfig(
            service_tier="flex",
            tools=[types.Tool(googleSearch=types.GoogleSearch())],
        ),
    )

    print("Full response object:")
    print(response)

    print("\nResponse text:")
    print(response.text)

    # Optional: Verify which inference tier actually served the request
    actual_tier = response.sdk_http_response.headers.get("x-gemini-service-tier")
    print(f"\nServed via tier: {actual_tier}")

except Exception as e:
    print(f"Flex request failed: {e}")
```

```python
# Extract token usage from the Gemini response metadata
usage = getattr(response, "usage_metadata", None)

input_token_count = getattr(usage, "prompt_token_count", 0) or 0
candidate_token_count = getattr(usage, "candidates_token_count", 0) or 0
thought_token_count = getattr(usage, "thoughts_token_count", 0) or 0
tool_token_count = getattr(usage, "tool_use_prompt_token_count", 0) or 0
total_token_count = getattr(usage, "total_token_count", 0) or 0

print("Input tokens:", input_token_count)
print("Candidate/output tokens:", candidate_token_count)
print("Thought tokens:", thought_token_count)
print("Tool-use prompt tokens:", tool_token_count)
print("Total tokens:", total_token_count)
```
