```python
import os

from google import genai
from google.genai import types


PROJECT_ID = os.getenv("PROJECT_ID") or os.getenv("GOOGLE_CLOUD_PROJECT")
LOCATION = os.getenv("VERTEX_AI_LOCATION", "global")
MODEL = "gemini-3-flash-preview"

if not PROJECT_ID:
    raise RuntimeError(
        "Set PROJECT_ID or GOOGLE_CLOUD_PROJECT before creating the Vertex AI client."
    )

# Authenticate with Application Default Credentials, for example:
# gcloud auth application-default login
client = genai.Client(
    vertexai=True,
    project=PROJECT_ID,
    location=LOCATION,
    http_options=types.HttpOptions(
        api_version="v1",
        timeout=600_000,
        headers={
            "X-Vertex-AI-LLM-Request-Type": "shared",
            "X-Vertex-AI-LLM-Shared-Request-Type": "flex",
        },
    ),
)

response = client.models.generate_content(
    model=MODEL,
    contents="What are the latest developments in Vertex AI? Cite your web sources.",
    config=types.GenerateContentConfig(
        temperature=0.2,
        # max_output_tokens=256,
        tools=[
            types.Tool(
                google_search=types.GoogleSearch(),
            )
        ],
    ),
)

print(response.text)
```
