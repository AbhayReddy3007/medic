```python
# Identify drug_names missing specific IP PDFs
import os
import pandas as pd

required_pdfs = [
    'Global Launch Sequencing & Arbitrage.pdf',
    'Patent Strength and Invalidity Opportunity.pdf',
    'Patent Thicket and Circumvention Feasibility.pdf',
    'Primary Market Entry Horizon.pdf',
]

# Use `result_df` if available, otherwise try to read `gcs_pdfs.csv`
if 'result_df' in globals():
    df = result_df.copy()
elif os.path.exists('gcs_pdfs.csv'):
    df = pd.read_csv('gcs_pdfs.csv')
else:
    raise FileNotFoundError('No `result_df` in memory and `gcs_pdfs.csv` not found.')

# Focus on IP pillar entries
if 'pillar' in df.columns:
    df_ip = df[df['pillar'].str.upper() == 'IP']
else:
    df_ip = df.copy()

# Build set of pdfs per drug
pdfs_by_drug = df_ip.groupby('drug_name')['pdf_name'].apply(set).reset_index()

# Compute missing PDFs
pdfs_by_drug['missing_pdfs'] = pdfs_by_drug['pdf_name'].apply(lambda s: sorted([p for p in required_pdfs if p not in s]))

# Filter drugs with at least one missing required PDF
missing_df = pdfs_by_drug[pdfs_by_drug['missing_pdfs'].map(len) > 0][['drug_name', 'missing_pdfs']]

print('Drugs missing one or more required IP PDFs:')
print(missing_df.to_string(index=False))

# Optionally save to CSV
missing_df.to_csv('missing_ip_pdfs_by_drug.csv', index=False)
print('\nSaved summary to missing_ip_pdfs_by_drug.csv')
```
